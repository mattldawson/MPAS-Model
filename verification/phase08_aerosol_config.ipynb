# Phase 8: Runtime Aerosol Representation Configuration

Verifies that MIAM aerosol representations are dynamically allocated
at runtime based on the MIAM config file:
- Aerosol fields appear in output NetCDF matching the MIAM config
- Switching configs (1 → 2 sections) changes fields without recompilation
- Gas-phase results unchanged from earlier phases

**Pre-requisite:**
- `data/jw_480km_aerosol_1sec/output.nc` (1-section config)
- `data/jw_480km_aerosol_2sec/output.nc` (2-section config)

```python
import netCDF4 as nc
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

DATA_DIR = Path("..") / "data"
SEC1 = DATA_DIR / "jw_480km_aerosol_1sec" / "output.nc"
SEC2 = DATA_DIR / "jw_480km_aerosol_2sec" / "output.nc"
```

## 1. Aerosol Fields Comparison

Compare variable sets between 1-section and 2-section configs.

```python
for label, path in [("1-section", SEC1), ("2-section", SEC2)]:
    if path.exists():
        ds = nc.Dataset(path)
        aerosol_vars = [v for v in ds.variables
                        if "aerosol" in v.lower() or "section" in v.lower()]
        print(f"{label}: {len(aerosol_vars)} aerosol fields — {aerosol_vars}")
        ds.close()
    else:
        print(f"{label}: file not found — Phase 8 not yet run")
```
